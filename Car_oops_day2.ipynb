```python
# Creating a Car object(display info,brake,accelerate)

class Car:
    #constructor
    def __init__(self,color,brand,model):
        self.color = color
        self.brand = brand
        self.model = model
        self.speed = 0
    
    #display_info()
    def display_info(self):
        print(f"Car color: {self.color}, Car brand: {self.brand}")
        print(f"Car model: {self.model}")
        print(f"Current Speed: {self.speed}")

    # accelerate 
    def accelerate(self,increase):
        self.speed += increase
        print(f"The {self.model} is accelerating with {self.speed}km/hr")

    #brake
    def brake(self,decrease):
        self.speed -=decrease
        if self.speed <0:
            self.speed = 0
        print(f"The {self.model} is deaccelerating with {self.speed} km/hr")

       

```

```python
car1 = Car("metallic","Tata","Tiago")
car2 = Car("red","Maruti","Dzire")
```

```python
car1.display_info()
```

```python
car1.accelerate(40)
```

```python
car1.brake(20)
```

```python
car2.display_info()
```

```python
car2.accelerate(70)
```

```python
car2.brake(20)
```
