```python
# LabelEncoder converts the text/categorical columns into numerical data

from sklearn.preprocessing import LabelEncoder
import pandas as pd

#creating data frames
data = pd.DataFrame({'Education':['High School','Bachelor','Master','PhD','Bachelor','High School']})
```

```python
data
```

```python
le = LabelEncoder()
le.fit_transform(data['Education'])
```

```python
print('classes:',le.classes_)
```

```python
print(le.inverse_transform([0,1,2,3]))
```

```python
# One Hot Encoding
import pandas as pd
from sklearn.preprocessing import OneHotEncoder
#Animal Type (Nominal - no order)
data =pd.DataFrame({'Animal':['Cat','Dog','Fish','Dog','Cat']})
data
```

```python
#Method1-pandas get_dummies()
encoded = pd.get_dummies(data, columns = ['Animal'])
print(encoded)
```

```python
#Method1-pandas get_dummies() dropping first column
encoded = pd.get_dummies(data, columns = ['Animal'],drop_first=True)
print(encoded)
```

```python
ohe =OneHotEncoder(drop="first",sparse_output= False)
result = ohe.fit_transform(data [["Animal"]])
col_name = ohe.get_feature_names_out(["Animal"])
encoded_df = pd.DataFrame(result, columns=col_name)
encoded_df
```

```python
col_name
encoded_df
```

```python
#ordinal encoding is like label encoding but define order

from sklearn.preprocessing import OrdinalEncoder
import pandas as pd
data = pd.DataFrame({'Size':['Small','Large','Medium','XL','Small','Medium'],
                     'Grade':['Poor','Good','Average','Excellent','Good','Average']})
oe = OrdinalEncoder(categories =[
    ["Small","Medium","Large","XL"],
    ["Poor","Average","Good","Excellent"]])
data[["size_enc","grade_enc"]] = oe.fit_transform(data[['Size','Grade']])
print(data)
```

```python
# Scaling method MInMaxScaler use in Deep Learning (Normalization)

from sklearn.preprocessing import MinMaxScaler
import pandas as pd
import numpy as np
data = pd.DataFrame({'Age':[25,30,35,40,45,50],
                      'Salary':[30000,45000,60000,75000,90000,120000]})
data
scaler = MinMaxScaler(feature_range= (0,1))
data_scaled = scaler.fit_transform(data)
df_scaled = pd.DataFrame( data_scaled,columns =['Age_scaled','Salary_scaled'])
print(df_scaled)
```
