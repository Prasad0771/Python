class Student add_subject,add_marks,calculate_avg,display_info

```python
class Student:
    def __init__(self,name,roll_no,grade):
        self.name = name
        self.roll_no = roll_no
        self.grade = grade
        self.subjects = []
        self.marks = {}
# method to add subjects
    def add_subjects(self,subject_name):
        if subject_name not in self.subjects:
            self.subjects.append(subject_name)
            print(f"subjects '{subject_name}' is added for {self.name}")
        else:
            print(f"{self.name} is already enrolled for {subject_name}")
# method to add marks    
    def add_marks(self,subject_name,marks):
        if subject_name in self.subjects:
            self.marks[subject_name] = marks
            print(f"marks updated:{subject_name} = {marks}")
            print(self.marks)
        else:
            print(f"Error:{subject_name} is not the enrolled subjects")
# methods to calculate average
    def calculate_average(self):
        if len(self.marks) ==0:
            return 0
        total = sum(self.marks.values())
        average = total/len(self.marks)
        return average
#method to display student_info
    def display_info(self):
        print(f"Student_name:'{self.name}'")
        print(f"Roll_no: '{self.roll_no}'")
        print(f"grade: {self.grade}")
        print(f"subjects enrolled:{','.join(self.subjects) if self.subjects else "none"}")
        if self.marks:
            print("-----Marks-----")
            for subject,mark in self.marks.items():
                print(f"--{subject} {mark}")
            print(f"Average:{self.calculate_average():.2f}")
        else:
            print("No marks recorded yet!")
# methos to check pass/fail
    def check_result(self):
        avg = self.calculate_average()
        if avg >= 40:
            print(f"{self.name:} is pass with {avg}%")
        else:
            print(f"{self.name} is failed with {avg}%")
        
```
