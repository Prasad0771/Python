Introduction of Python Object oriented programming.

```python
class BankAccount:
    # constructor
    def __init__(self,acc_holder,acc_number,balance):
     self.acc_holder = acc_holder
     self.acc_number = acc_number
     self.balance = balance

     #Deposting the amount
    def deposit(self,amount):
     self.balance += amount
     print(f"Deposited:{amount}, New Balance :{self.balance}")

    #withdraw th money
    def withdraw(self, amount):
        if self.balance >= amount:
            self.balance -=  amount
            print(f"Withdrawn:{amount}, New Balance :{self.balance}")
        else:
            print("Insufficient Fund")  
    
    #Check Balance
    def checkbalance(self):
        print(f"current balance is : {self.balance}")
```

```python
Account1 = BankAccount("Abani", "Acc001",5000)
Account2 = BankAccount("Julie","Acc002",7000)
```

```python
Account1.deposit(1000)
```

```python
Account1.withdraw(500)
```

```python
Account1.checkbalance()
```

```python
Account2.checkbalance()
```

```python
Account2.deposit(700)
```
