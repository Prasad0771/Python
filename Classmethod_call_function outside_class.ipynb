```python
#Programme to method call function defined outside the class
```

```python
def Scale_10(x):
    return x*10

class ABC:
    def __init__(self, var):self.var = var
    def display_info(self):
        print("var is:",self.var)

    def modify(self):
        self.var = Scale_10(self.var)
```

```python
ABC = ABC(16)
```

```python
ABC.display_info()
```

```python
ABC.modify()
```

```python
ABC.display_info()
```
